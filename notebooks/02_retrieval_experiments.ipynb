# 02 Retrieval Experiments

Compare TF-IDF, BM25, dense, and hybrid retrieval.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(ROOT / "src"))
from retrieve_tfidf import TFIDFRetriever
from evaluate_retrieval import evaluate_retriever

claims = pd.read_csv(ROOT / "data" / "processed" / "claims.csv")
passages = pd.read_csv(ROOT / "data" / "processed" / "evidence_passages.csv")
claims_small = claims.head(25)
retriever = TFIDFRetriever(passages)
evaluate_retriever("tfidf", retriever, claims_small, top_k=10)
```

```python
# Full comparison can be run from the terminal:
# python src/evaluate_retrieval.py --max_claims 100
results_path = ROOT / "data" / "processed" / "retrieval_results.csv"
if results_path.exists():
    results = pd.read_csv(results_path)
    display(results)
    if "recall@5" in results:
        results.plot(kind="bar", x="method", y="recall@5", title="Recall@5 by retrieval method")
```
