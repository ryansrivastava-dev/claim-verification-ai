# 01 Dataset Exploration

Explore loaded SciFact claims, labels, and evidence passages. Run `python src/load_data.py` before this notebook.

```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(ROOT / "src"))
processed = ROOT / "data" / "processed"
claims_path = processed / "claims.csv"
passages_path = processed / "evidence_passages.csv"

if not claims_path.exists():
    raise FileNotFoundError("Run python src/load_data.py first.")

claims = pd.read_csv(claims_path)
passages = pd.read_csv(passages_path) if passages_path.exists() else pd.DataFrame()
claims.head()
```

```python
print("Claims:", len(claims))
print("Evidence passages:", len(passages))
print("Columns:", claims.columns.tolist())
claims["label"].value_counts(dropna=False)
```

```python
if "label" in claims.columns:
    ax = claims["label"].value_counts().plot(kind="bar", title="Label distribution")
    ax.set_xlabel("Label")
    ax.set_ylabel("Count")
    plt.tight_layout()
```

```python
claims[["claim", "label", "evidence_doc_ids"]].sample(min(5, len(claims)), random_state=42)
```
