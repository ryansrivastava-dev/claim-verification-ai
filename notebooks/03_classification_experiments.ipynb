# 03 Classification Experiments

Train and compare lightweight claim verification classifiers.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(ROOT / "src"))
from train_classifier import load_training_data, split_data, train_tfidf_logreg, evaluate_model

df = load_training_data(ROOT / "data" / "processed" / "claim_evidence_pairs.csv")
X_train, X_test, y_train, y_test = split_data(df)
model = train_tfidf_logreg(X_train, y_train)
evaluate_model(model, X_test, y_test)
```
